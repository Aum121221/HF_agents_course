```python
from getpass import getpass
from smolagents import CodeAgent, OpenAIModel, tool

OPENROUTER_API_KEY = getpass("Enter your OpenRouter API key: ")

@tool
def catering_service_tool(query: str) -> str:
    """
    This tool returns the highest-rated catering service in Gotham City.

    Args:
        query: A search term for finding catering services.
    """
    services = {
        "Gotham Catering Co.": 4.9,
        "Wayne Manor Catering": 4.8,
        "Gotham City Events": 4.7,
    }

    best_service = max(services, key=services.get)
    return best_service


model = OpenAIModel(
    model_id="openrouter/free",
    api_base="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)

agent = CodeAgent(tools=[catering_service_tool], model=model)

result = agent.run(
    "Can you give me the name of the highest-rated catering service in Gotham City?"
)

print(result)
```

```python
from smolagents import Tool, CodeAgent, OpenAIModel

class SuperheroPartyThemeTool(Tool):
    name = "superhero_party_theme_generator"
    description = """
    This tool suggests creative superhero-themed party ideas based on a category.
    It returns a unique party theme idea."""

    inputs = {
        "category": {
            "type": "string",
            "description": "The type of superhero party (e.g., 'classic heroes', 'villain masquerade', 'futuristic Gotham').",
        }
    }

    output_type = "string"

    def forward(self, category: str):
        themes = {
            "classic heroes": "Justice League Gala: Guests come dressed as their favorite DC heroes with themed cocktails like 'The Kryptonite Punch'.",
            "villain masquerade": "Gotham Rogues' Ball: A mysterious masquerade where guests dress as classic Batman villains.",
            "futuristic Gotham": "Neo-Gotham Night: A cyberpunk-style party inspired by Batman Beyond, with neon decorations and futuristic gadgets."
        }

        return themes.get(category.lower(), "Themed party idea not found. Try 'classic heroes', 'villain masquerade', or 'futuristic Gotham'.")


party_theme_tool = SuperheroPartyThemeTool()

model = OpenAIModel(
    model_id="openrouter/free", # Changed model_id to 'openrouter/free'
    api_base="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)

agent = CodeAgent(tools=[party_theme_tool], model=model)

result = agent.run(
    "What would be a good superhero party idea for a 'villain masquerade' theme?"
)

print(result)
```

```python
import requests
from smolagents import ToolCollection, CodeAgent, InferenceClientModel
from smolagents.tools import BaseTool # Changed back to BaseTool for compatibility
from huggingface_hub import login

# Call login to authenticate with Hugging Face
login()

# Helper function for PubMed search (can be defined outside the class)
def _perform_pubmed_search(query: str):
    """Search PubMed articles via NCBI E-utilities API"""
    url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"
    params = {"db": "pubmed", "term": query, "retmode": "json", "retmax": 5}
    response = requests.get(url, params=params).json()
    ids = response.get("esearchresult", {}).get("idlist", [])
    return {"query": query, "pubmed_ids": ids}

# Define PubMedTool as a subclass of BaseTool to ensure compatibility with CodeAgent
class PubMedTool(BaseTool):
    name = "pubmed_search"
    description = "Search PubMed articles via NCBI E-utilities API"

    # Required by BaseTool, this is the method that the agent will call
    def __call__(self, query: str):
        return _perform_pubmed_search(query)

    # Required by CodeAgent for internal prompt rendering
    def to_code_prompt(self) -> str:
        return f"""
# Tool: {self.name}
# Description: {self.description}
# Usage:
# result = {self.name}(query="search terms")
# Example:
# result = pubmed_search(query="remedy for hangover")
"""

# ✅ Instantiate the PubMedTool
pubmed_instance = PubMedTool()

# ✅ Load into ToolCollection
tool_collection = ToolCollection(tools=[pubmed_instance])

# ✅ Use Hugging Face model for reasoning
model = InferenceClientModel("Qwen/Qwen2.5-Coder-32B-Instruct")

# ✅ Create agent with PubMed tool + base tools
# The tools argument expects a list of BaseTool instances.
agent = CodeAgent(tools=[*tool_collection.tools], model=model, add_base_tools=True)

# ✅ Run a query
result = agent.run("Please find a remedy for hangover.")
print(result)
```

```python
from getpass import getpass
from huggingface_hub import InferenceClient
from smolagents import CodeAgent, LiteLLMModel, tool

hf_token = HF_TOKEN

image_client = InferenceClient(token=hf_token)

@tool
def image_generator(prompt: str) -> str:
    """
    Generate an image from a text prompt and save it locally.

    Args:
        prompt: The detailed image prompt to generate from.

    Returns:
        The local file path of the generated image.
    """
    image = image_client.text_to_image(
        prompt,
        model="black-forest-labs/FLUX.1-schnell",
    )
    path = "generated_image.png"
    image.save(path)
    return path

model = LiteLLMModel(
    model_id="huggingface/Qwen/Qwen2.5-Coder-32B-Instruct",
    api_key=hf_token,
)

agent = CodeAgent(
    tools=[image_generator],
    model=model,
    max_steps=4,
)

agent.run(
    """
Improve user_prompt, then generate an image with image_generator.

Do not add Midjourney parameters such as --v, --ar, or --style.
Return only the generated image path and the improved prompt.
""",
    additional_args={
        "user_prompt": "A grand superhero-themed party at Wayne Manor, with Alfred overseeing a luxurious gala"
    },
)
```

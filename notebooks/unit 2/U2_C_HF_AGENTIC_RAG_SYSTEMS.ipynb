````python
from langchain_community.docstore.document import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.retrievers import BM25Retriever

from smolagents import Tool, CodeAgent, OpenAIModel
````

````python
model = OpenAIModel(
    model_id="openrouter/free",
    api_base="https://openrouter.ai/api/v1",
    api_key=OPENROUTER_API_KEY,
)
````

````python
class PartyPlanningRetrieverTool(Tool):
    name = "party_planning_retriever"
    description = (
        "Uses search to retrieve relevant high-end superhero-themed "
        "party planning ideas, including decorations, entertainment, and catering."
    )

    inputs = {
        "query": {
            "type": "string",
            "description": (
                "The query to perform. This should be related to luxury superhero "
                "party planning, decorations, entertainment, catering, or immersive experiences."
            ),
        }
    }

    output_type = "string"

    def __init__(self, docs, **kwargs):
        super().__init__(**kwargs)
        self.retriever = BM25Retriever.from_documents(
            docs,
            k=5,
        )

    def forward(self, query: str) -> str:
        assert isinstance(query, str), "Your search query must be a string"

        docs = self.retriever.invoke(query)

        return "\nRetrieved ideas:\n" + "".join(
            [
                f"\n\n===== Idea {i} =====\n"
                f"Source: {doc.metadata.get('source', 'Unknown')}\n"
                f"{doc.page_content}"
                for i, doc in enumerate(docs, start=1)
            ]
        )
````

````python
party_ideas = [
    {
        "text": (
            "Decorations - The Heroic Environment: Use immersive backdrops and photo booths "
            "instead of simple banners. Include high-resolution, life-sized character backdrops "
            "such as a Gotham City skyline or an Avengers Stark Tower lounge. Add professional "
            "lighting, ring lights, neon accents, and high-quality character props for a red-carpet feel."
        ),
        "source": "Decorations - Immersive Backdrops & Photo Booths",
    },
    {
        "text": (
            "Decorations - Premium Tableware and Linens: Move beyond paper plates to custom-designed, "
            "heavy-duty themed ceramic sets or high-end metallic-finish disposables in primary colors "
            "such as Deep Red, Royal Blue, and Gold. Use themed runners such as web lace for Spider-Man "
            "or tech-circuit patterns for Iron Man."
        ),
        "source": "Decorations - Premium Tableware & Linens",
    },
    {
        "text": (
            "Decorations - Atmospheric Lighting and Balloons: Use professional balloon installations, "
            "organic balloon arches in metallic superhero colors, and LED-lit decor. Consider holographic "
            "or neon lighting to mimic the high-tech feel of superhero headquarters."
        ),
        "source": "Decorations - Atmospheric Lighting & Balloons",
    },
    {
        "text": (
            "Entertainment - Elite Character Performers: Hire professional, screen-accurate actors who "
            "can interact with guests in character rather than just posing for photos. This adds "
            "sophistication and immersion."
        ),
        "source": "Entertainment - Elite Character Performers",
    },
    {
        "text": (
            "Entertainment - Interactive Tech and Games: Create a VR Training Academy where guests can "
            "experience superpower training through Virtual Reality. Add heroic competitions such as "
            "a high-stakes Mystery Villain scavenger hunt."
        ),
        "source": "Entertainment - Interactive Tech & Games",
    },
    {
        "text": (
            "Entertainment - Themed Music and Atmosphere: Hire a professional DJ to play cinematic "
            "orchestral soundtracks in a Hans Zimmer style mixed with modern hits to maintain a "
            "high-energy luxury vibe."
        ),
        "source": "Entertainment - Themed Music & Atmosphere",
    },
    {
        "text": (
            "Catering - Curated Themed Menus: Work with a luxury caterer to design a menu where each "
            "course is inspired by a hero or villain, such as Kryptonite green-tinted appetizers, "
            "Wakanda-inspired exotic fruit and spice platters, or Stark Industries molecular gastronomy bites."
        ),
        "source": "Catering - Curated Themed Menus",
    },
    {
        "text": (
            "Catering - Signature Cocktails and Mocktails: Serve Super-Serum cocktails or mocktails "
            "in elegant glassware with dry ice for a dramatic smoky effect."
        ),
        "source": "Catering - Signature Cocktails & Mocktails",
    },
    {
        "text": (
            "Catering - Artisanal Dessert Displays: Create a grand dessert station featuring "
            "hyper-realistic superhero-themed cakes, such as a Captain America shield cake made of "
            "gourmet chocolate, along with themed macarons or petit fours."
        ),
        "source": "Catering - Artisanal Dessert Displays",
    },
]
````

````python
source_docs = [
    Document(
        page_content=doc["text"],
        metadata={"source": doc["source"]},
    )
    for doc in party_ideas
]

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    add_start_index=True,
    strip_whitespace=True,
    separators=["\n\n", "\n", ".", " ", ""],
)

docs_processed = text_splitter.split_documents(source_docs)

party_planning_retriever = PartyPlanningRetrieverTool(docs_processed)
````

````python
agent = CodeAgent(
    tools=[party_planning_retriever],
    model=model,
)
````

````python
response = agent.run(
    "Find ideas for a luxury superhero-themed party, including entertainment, catering, and decoration options."
)

print(response)
````

====================================================================================================
                        COMPLETE AGENT & RETRIEVAL (RAG) PROCESS SUMMARY
====================================================================================================

1. OVERVIEW: CODE 1 vs. CODE 2
----------------------------------------------------------------------------------------------------
• Code 1 (Live Web Search):
  - Uses `DuckDuckGoSearchTool` to query the live internet in real time.
  - Data comes dynamically from external websites across the web.

• Code 2 (Local RAG Pipeline):
  - Creates a custom local tool (`PartyPlanningRetrieverTool`) using `smolagents` and `LangChain`.
  - Restricts search strictly to a pre-defined local list of party planning ideas (`party_ideas`).
  - Does not query the external internet, keeping retrieval fast, reproducible, and private.


2. STEP-BY-STEP EXECUTION FLOW OF CODE 2 (LOCAL RAG)
----------------------------------------------------------------------------------------------------
  [ Step 1: Raw Data Creation ]
    └── Define `party_ideas`: A Python list of dictionaries containing text and source tags.

  [ Step 2: LangChain Document Conversion ]
    └── Convert dictionaries into LangChain `Document(page_content=..., metadata=...)` objects.

  [ Step 3: Text Splitting / Chunking ]
    └── Pass documents into `RecursiveCharacterTextSplitter`:
        - `chunk_size=500` characters
        - `chunk_overlap=50` characters
        - Keeps text units focused for precise retrieval.

  [ Step 4: Indexing with BM25 Retriever ]
    └── Pass chunks into `BM25Retriever.from_documents(docs_processed, k=5)`:
        - BM25 calculates relevance scores using Term Frequency (TF), Inverse Document
          Frequency (IDF), and document length normalization.
        - High-weighting goes to rare, specific keywords (e.g., "Kryptonite", "Gotham").
        - Low-weighting goes to common stop words ("the", "and").

  [ Step 5: Agent Interaction & Top-k Filtering ]
    └── User sends prompt to `CodeAgent`.
    └── Agent invokes `PartyPlanningRetrieverTool.forward(query)`.
    └── `BM25Retriever.invoke(query)` ranks all chunks by score and selects the TOP 5 (`k=5`).
    └── Chunks are formatted into a string and returned to the LLM agent to construct the final answer.


3. ARE THE SPLIT CHUNKS / QUERIES MEANINGFUL?
----------------------------------------------------------------------------------------------------
• Meaningfulness of Queries (Sent by Agent to Retriever):
  - YES. The LLM agent generates natural language search queries (e.g., "luxury superhero party
    catering ideas"). Because BM25 extracts key words from these queries, they are highly
    effective at surfacing relevant chunks.

• Meaningfulness of Split Chunks (Created by Text Splitter):
  - YES, BUT DEPENDS ON PARAMETERS:
    a) In this code (`chunk_size=500`, `chunk_overlap=50`), the chunks remain MEANINGFUL because
       500 characters roughly equal 75–100 words (a complete paragraph or standalone idea).
    b) The 50-character overlap ensures that context at chunk boundaries is not lost mid-sentence.
    c) If `chunk_size` were too small (e.g., 20 characters), chunks would become broken words
       and lose meaning.
    d) Because BM25 is keyword-based, semantic context isn't parsed deeply (like with vector
       embeddings), but 500-character chunks provide rich enough text for the LLM to understand
       the retrieved content completely.
